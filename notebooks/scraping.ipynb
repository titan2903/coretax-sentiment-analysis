# CoreTax (M-Pajak) Max Review Scraper
This notebook extracts all available reviews for the DJP (CoreTax) mobile app, merging results from multiple languages to maximize the data count.

```python
from google_play_scraper import Sort, reviews_all
import pandas as pd
import os

# App ID from the URL
APP_ID = 'id.go.pajak.djp'

# Path adjusted for notebook being inside 'scripts' folder
OUTPUT_FILE = '../data/raw/coretax_reviews.csv'

# Create data directory if not exists (one level up from 'scripts')
if not os.path.exists('../data'):
    os.makedirs('../data')
```

```python
print(f"Fetching reviews for: {APP_ID}...")

# 1. Fetching Indonesian reviews
print("[*] Collecting Indonesian (id) reviews...")
res_id = reviews_all(
    APP_ID,
    lang='id',
    country='id',
    sort=Sort.NEWEST
)

# 2. Fetching English reviews (to catch users with English phone settings)
print("[*] Collecting English (en) reviews...")
res_en = reviews_all(
    APP_ID,
    lang='en',
    country='id',
    sort=Sort.NEWEST
)

# Merge and remove duplicates based on reviewId
all_reviews = res_id + res_en
df = pd.DataFrame(all_reviews)
if not df.empty:
    df.drop_duplicates(subset=['reviewId'], inplace=True)
    print(f"[✓] Successfully collected {len(df)} unique text reviews.")
else:
    print("[!] No reviews found.")
```

```python
# Final processing and saving
if not df.empty:
    # Select relevant columns
    df_final = df[['userName', 'score', 'at', 'content', 'thumbsUpCount', 'reviewCreatedVersion']]
    df_final.columns = ['user_name', 'rating', 'date', 'review_text', 'likes', 'app_version']
    
    # Save to CSV
    df_final.to_csv(OUTPUT_FILE, index=False, encoding='utf-8-sig')
    print(f"[✓] Results saved to: {OUTPUT_FILE}")
    
    print("\nNote: If the count is still below the Play Store total (~13k), ")
    print("the remaining are likely ratings WITHOUT any review text.")
    display(df_final.head())
else:
    print("No data to process.")
```
